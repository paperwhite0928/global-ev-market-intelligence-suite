# 05 — Strategic Outcomes

Scenario framing: hybrid-led transition vs accelerated BEV pivot trade-offs.

```python
import sys
from pathlib import Path

import pandas as pd

ROOT = Path.cwd().parent if Path.cwd().name == "notebooks" else Path.cwd()
sys.path.insert(0, str(ROOT))

from src.load_data import load_processed

mix = load_processed("toyota_mix")
fin = load_processed("financials")
bev = load_processed("bev_sales")

global_2025 = mix[(mix["region"] == "Global") & (mix["year"] == 2025)]
total = global_2025["units"].sum()
bev_share = global_2025[global_2025["powertrain"] == "BEV"]["units"].sum() / total * 100
toyota_bev_share = bev[(bev["year"] == 2025) & (bev["company"] == "Toyota")]["market_share_pct"].iloc[0]

print(f"Toyota global electrified BEV mix: {bev_share:.1f}%")
print(f"Toyota global BEV market share: {toyota_bev_share:.1f}%")
print(f"Toyota operating margin (2025): {fin.iloc[-1]['operating_margin_pct']:.1f}%")
```
