Analyze Toyota's electrified sales mix and hybrid-first positioning vs rising BEV share.

```python
import sys
from pathlib import Path

import matplotlib.pyplot as plt
import pandas as pd

ROOT = Path.cwd().parent if Path.cwd().name == "notebooks" else Path.cwd()
sys.path.insert(0, str(ROOT))

from src.load_data import load_processed
from src.visualization import plot_toyota_mix_stacked

mix = load_processed("toyota_mix")
mix.head()
```

```python
global_mix = mix[mix["region"] == "Global"].pivot_table(
    index="year", columns="powertrain", values="units", aggfunc="sum", fill_value=0
)
global_mix["bev_share_pct"] = global_mix["BEV"] / global_mix.sum(axis=1) * 100
global_mix[["HEV", "PHEV", "BEV", "FCEV", "bev_share_pct"]]
```
