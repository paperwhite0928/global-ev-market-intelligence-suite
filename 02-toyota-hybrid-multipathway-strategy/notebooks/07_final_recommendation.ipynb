```python
import sys
from pathlib import Path

import pandas as pd

ROOT = Path.cwd().parent if Path.cwd().name == "notebooks" else Path.cwd()
sys.path.insert(0, str(ROOT))

from src.load_data import load_processed

mix = load_processed("toyota_mix")
bev = load_processed("bev_sales")
fin = load_processed("financials")
risk = load_processed("risk_summary")
battery = load_processed("battery_index")

latest_year = int(mix["year"].max())
global_mix = mix[(mix["region"] == "Global") & (mix["year"] == latest_year)]
total_units = global_mix["units"].sum()
bev_units = global_mix.loc[global_mix["powertrain"] == "BEV", "units"].sum()
hev_units = global_mix.loc[global_mix["powertrain"] == "HEV", "units"].sum()

toyota_bev = bev[(bev["year"] == latest_year) & (bev["company"] == "Toyota")]
peer_bev = bev[bev["year"] == latest_year].sort_values("bev_units", ascending=False)

summary = pd.DataFrame([
    {"metric": f"Toyota global BEV mix ({latest_year})", "value": f"{bev_units / total_units * 100:.1f}%"},
    {"metric": f"Toyota global HEV mix ({latest_year})", "value": f"{hev_units / total_units * 100:.1f}%"},
    {"metric": f"Toyota BEV market share ({latest_year})", "value": f"{toyota_bev['market_share_pct'].iloc[0]:.1f}%" if len(toyota_bev) else "N/A"},
    {"metric": f"Toyota operating margin ({latest_year})", "value": f"{fin[fin['year'] == latest_year]['operating_margin_pct'].iloc[0]:.1f}%"},
    {"metric": "Battery cost index (latest)", "value": f"{battery.iloc[-1]['composite_index']:.1f}"},
    {"metric": "Highest regional BEV risk", "value": risk.iloc[0]['region']},
])
summary
```

## 4. 2030 strategy comparison

Score each pathway using your updated data. Adjust weights and targets as needed.

```python
scenarios = pd.DataFrame([
    {
        "strategy": "Hybrid-led (status quo)",
        "bev_mix_2030_pct": 12,
        "capex_usd_b": 6,
        "margin_impact_pp": 0.0,
        "regulatory_risk": "High in EU",
        "share_risk": "High vs BYD/Tesla",
    },
    {
        "strategy": "Balanced pivot",
        "bev_mix_2030_pct": 25,
        "capex_usd_b": 12,
        "margin_impact_pp": -1.2,
        "regulatory_risk": "Moderate",
        "share_risk": "Moderate",
    },
    {
        "strategy": "Aggressive BEV",
        "bev_mix_2030_pct": 40,
        "capex_usd_b": 20,
        "margin_impact_pp": -2.5,
        "regulatory_risk": "Low in EU/China",
        "share_risk": "Lower if scale achieved",
    },
])

# Simple score: lower is better for cost/risk trade-off (customize after real data)
scenarios["composite_score"] = (
    scenarios["capex_usd_b"] * 0.3
    + abs(scenarios["margin_impact_pp"]) * 2
    + (40 - scenarios["bev_mix_2030_pct"]) * 0.15  # penalty for low BEV mix by 2030
)
scenarios.sort_values("composite_score")
```
